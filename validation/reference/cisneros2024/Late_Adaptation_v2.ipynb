```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math 
import scipy.stats as stats
```

```python
#95 percent CI given by the error bars of graph converted to standard deviation
def CI_95_to_SD(CI,N):
    #CI = 1.96 * SE
    #SE = CI/1.96
    #SE = SD/root(Sample Size)
    #SD = SE * root(Sample Size)
     
    SE = CI/1.96
    
    SD = SE * np.sqrt(N)
    
    return SD

def SE_to_SD(SE, N):
    return SE / np.sqrt(N)
```

```python
#get cohen's d
def SMD(mean1, mean2, sd1, sd2, n1, n2):
    
    diff_means = (mean1 - mean2)
    pooled_sd = np.sqrt(((n1 - 1) * (sd1**2) + (n2 - 1) * (sd2**2))/ (n1 + n2 - 2))
    
    return diff_means / pooled_sd

# compute effect size standard error 
def SE_SMD(n1,n2,d):
    
    return np.sqrt((n1 + n2)/(n1 * n2) + d**2 / (2 * (n1 + n2 -2)))


def get_95_CI(d, SE_coh):
    low = d - 1.96 * SE_coh
    high = d + 1.96 * SE_coh
    
    return low, high
```
